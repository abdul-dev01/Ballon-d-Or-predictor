```python
import unicodedata

def clean_name(name):
    name = unicodedata.normalize("NFKD", name).encode("ascii", "ignore").decode("utf-8")
    return name.strip().lower()

full_df["player_clean"] = full_df["player"].apply(clean_name)
nominees_clean = [clean_name(n) for n in nominees]

found = full_df[full_df["player_clean"].isin(nominees_clean)]["player"].unique()
missing = [n for n in nominees if clean_name(n) not in [clean_name(f) for f in found]]
print(missing)
print(len(found), "of", len(nominees), "nominees found")
```

```python
trophies = {
    "Jude Bellingham": 0,
    "Pau Cubarsí": 2,          # World Cup + La Liga
    "Marc Cucurella": 1,       # World Cup
    "Ousmane Dembélé": 2,      # Champions League + Ligue 1
    "Luis Díaz": 2,            # Bundesliga + German Cup
    "Bruno Fernandes": 0,
    "Gabriel": 1,              # Premier League
    "Erling Haaland": 2,       # FA Cup + EFL Cup
    "Achraf Hakimi": 3,        # Champions League + Ligue 1 + AFCON
    "Harry Kane": 2,           # Bundesliga + German Cup
    "Khvicha Kvaratskhelia": 2, # Champions League + Ligue 1
    "Lamine Yamal": 2,         # World Cup + La Liga
    "Sadio Mané": 1,           # Saudi Pro League
    "Marquinhos": 2,           # Champions League + Ligue 1
    "Lautaro Martínez": 2,     # Serie A + Italian Cup
    "Kylian Mbappé": 0,
    "Nuno Mendes": 2,          # Champions League + Ligue 1
    "Lionel Messi": 1,         # MLS Cup
    "João Neves": 2,           # Champions League + Ligue 1
    "Michael Olise": 2,        # Bundesliga + German Cup
    "Willian Pacho": 2,        # Champions League + Ligue 1
    "Julián Quiñones": 0,
    "Declan Rice": 1,          # Premier League
    "Rodri": 3,                # World Cup + FA Cup + EFL Cup
    "Fabián Ruiz": 3,          # World Cup + Champions League + Ligue 1
    "William Saliba": 1,       # Premier League
    "Ferran Torres": 2,        # World Cup + La Liga
    "Dayot Upamecano": 2,      # Bundesliga + German Cup
    "Vinícius Júnior": 0,
    "Vitinha": 2,              # Champions League + Ligue 1
}

nominee_df["trophies_final"] = nominee_df["player"].map(trophies).fillna(0)
```

```python
nominee_df["score"] = (
    nominee_df["goals_final"] * 1.0 +
    nominee_df["assists_final"] * 0.7 +
    nominee_df["trophies_final"] * 5.0
)

nominee_df["probability"] = (nominee_df["score"] / nominee_df["score"].sum() * 100).round(1)
nominee_df = nominee_df.sort_values("probability", ascending=False)

print(nominee_df[["player", "score", "probability"]])
```

```python
import json

output_df = nominee_df.rename(columns={"player": "name", "team": "club"})
output = output_df[["name", "club", "probability"]].to_dict(orient="records")

with open("nominees.json", "w", encoding="utf-8") as f:
    json.dump(output, f, indent=2, ensure_ascii=False)
```
